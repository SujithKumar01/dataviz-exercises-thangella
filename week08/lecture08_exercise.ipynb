## Task 1 — World choropleth: life expectancy diverging scale

**What to build:** A world choropleth showing **life expectancy relative to the global average** using a diverging colour scale.

**Requirements:**
- Use the Gapminder dataset for 2007: `px.data.gapminder()`
- Compute each country's deviation from the global mean life expectancy
- Diverging scale centred at zero (= world average)
- `hover_data` showing country name, raw life expectancy, and deviation
- Insight title naming which region is furthest below average

> 💡 `gm_2007['lifeExp'].mean()` gives you the global average to subtract from


````python
# Task 1
# YOUR CODE HERE
# Task 1 — World choropleth: life expectancy relative to global average

gm = px.data.gapminder()
gm_2007 = gm[gm['year'] == 2007].copy()

world_avg = gm_2007['lifeExp'].mean()

gm_2007['LifeExp_vs_Avg'] = gm_2007['lifeExp'] - world_avg

fig = px.choropleth(
    gm_2007,
    locations='iso_alpha',
    color='LifeExp_vs_Avg',
    hover_name='country',
    color_continuous_scale='RdBu',
    color_continuous_midpoint=0,
    labels={'LifeExp_vs_Avg': 'Years vs Global Avg'},
    projection='natural earth',
    height=600
)

fig.update_layout(
    title=f'Life expectancy varies by more than 30 years across countries relative to the global average of {world_avg:.1f} years',
    font=dict(family='Arial', size=12),
    paper_bgcolor='white',
    margin=dict(l=10, r=10, t=60, b=10),
    coloraxis_colorbar=dict(
        title='Years vs<br>Global Avg'
    )
)

# Annotate country with highest life expectancy
best_country = gm_2007.loc[gm_2007['lifeExp'].idxmax()]

fig.add_annotation(
    x=0.02,
    y=0.02,
    xref='paper',
    yref='paper',
    text=f"Highest: {best_country['country']} ({best_country['lifeExp']:.1f} years)",
    showarrow=False,
    bgcolor='white',
    bordercolor='black',
    borderwidth=1
)

fig.show()
````
